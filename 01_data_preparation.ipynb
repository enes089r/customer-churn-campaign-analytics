```python
import kagglehub
import pandas as pd

path = kagglehub.dataset_download("frtgnn/dunnhumby-the-complete-journey")

print("Path to dataset files:", path)
```

```python
import os

files = os.listdir(path)
print(files)
```

```python
transactions = pd.read_csv(f"{path}/transaction_data.csv")
demographics = pd.read_csv(f"{path}/hh_demographic.csv")
campaign_desc = pd.read_csv(f"{path}/campaign_desc.csv")
campaign_table = pd.read_csv(f"{path}/campaign_table.csv")
coupon = pd.read_csv(f"{path}/coupon.csv")
coupon_redempt = pd.read_csv(f"{path}/coupon_redempt.csv")
product = pd.read_csv(f"{path}/product.csv")

print(transactions.shape)
print(transactions.head())
```

```python
transactions.columns = transactions.columns.str.lower()
print(transactions.columns.tolist())
```

```python
print(demographics.shape)
print(campaign_desc.shape)
print(campaign_table.shape)
print(coupon.shape)
print(coupon_redempt.shape)
print(product.shape)
```

```python
print("Number of unique households:", transactions['household_key'].nunique())
print("Number of unique products:", transactions['product_id'].nunique())
print("Total revenue:", transactions['sales_value'].sum())
print("First week:", transactions['week_no'].min())
print("Last week:", transactions['week_no'].max())
```

```python
household_summary = transactions.groupby('household_key').agg(
    total_spend=('sales_value', 'sum'),
    num_transactions=('basket_id', 'nunique'),
    num_products=('product_id', 'nunique'),
    last_purchase_week=('week_no', 'max'),
    first_purchase_week=('week_no', 'min')
)

print(household_summary.shape)
household_summary.head()
```
